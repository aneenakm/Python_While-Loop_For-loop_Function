## Number Guessing Game

```python
import random 
secret_num =random.randint(1,10)
attempts =3
while attempts > 0:
    guess= int(input("Guess the number (between 1 and 10): "))
    if guess  < 1 or guess > 10 :
        print("Invalid Guess!!! choose between 1 to 10")
        continue
    attempts -= 1
    if guess > secret_num :
        print("Oops.. Try to guess a lower number")
    elif guess < secret_num :
        print("Oops.. Try to guess a higher number")
    else:
        print("Congratulation Wohoo!!,You Won..")
        break
else :
    print("Better luck next time!")
    print("The secret number was:", secret_num)
print("Game Ended!! Play Again")                        

```

## Multiplication Table Generator

```python
mul_num= int(input("Enter the number you want to multiplication table:"))
for i in range(1,11,1):
    print(mul_num,"x",i,"=", mul_num*i)
```

## BMI Calculator

```python
weight=float(input("Enter your weight in kg:"))
height=float(input("Enter your height in meter:"))
def calculate_bmi(weight, height):
        BMI = weight / (height * height)
        return BMI
if height <=0:
        print("Invalid Height! Please Enter a valid Height")
elif weight <=0:
    print("Invalid Weight! Please Enter a valid Weight")
else:
    res = calculate_bmi(weight,height)
    print("Your BMI is",format(res,".2f"))
```
